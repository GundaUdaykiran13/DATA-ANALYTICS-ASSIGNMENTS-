### Handle Duplicates

# Metric for Store Revenue Efficiency

**Metric**: Revenue per Square Meter

This metric measures how much revenue a store generates for each square meter of its physical space. It is calculated as:

`Revenue per Square Meter = Total Revenue (USD) / Square Meters`

A higher 'Revenue per Square Meter' indicates greater efficiency in utilizing the store's physical space to generate sales.

````python
total_revenue_by_store = sales_full.groupby('StoreKey')['Revenue_USD'].sum().reset_index()

# Merge with stores to get Square Meters
revenue_efficiency = total_revenue_by_store.merge(
    stores[['StoreKey', 'Square Meters']],
    on='StoreKey',
    how='left'
)

# Calculate Revenue per Square Meter
revenue_efficiency['Revenue_per_Square_Meter'] = (
    revenue_efficiency['Revenue_USD'] / revenue_efficiency['Square Meters']
)

# Sort and display the top 5 stores by this metric
top_stores_by_efficiency = revenue_efficiency.sort_values(
    by='Revenue_per_Square_Meter', ascending=False
).head(5)

print('Top 5 Stores by Revenue per Square Meter (USD):')
display(top_stores_by_efficiency)
````

# Metric for Revenue Growth

**Metric**: Month-on-Month Revenue Growth Rate

This metric measures the percentage change in revenue from one month to the next for each store. It is calculated as:

`Month-on-Month Revenue Growth Rate = ((Current Month Revenue - Previous Month Revenue) / Previous Month Revenue) * 100`

A positive growth rate indicates an increase in revenue, while a negative rate indicates a decrease.

````python
# Extract year and month from Order Date
sales_full['Order_YearMonth'] = sales_full['Order Date'].dt.to_period('M')

# Calculate monthly revenue for each store
monthly_revenue = sales_full.groupby(['StoreKey', 'Order_YearMonth'])['Revenue_USD'].sum().reset_index()

# Sort by StoreKey and Order_YearMonth to ensure correct lag calculation
monthly_revenue = monthly_revenue.sort_values(by=['StoreKey', 'Order_YearMonth'])

# Calculate previous month's revenue for each store
monthly_revenue['Previous_Month_Revenue'] = monthly_revenue.groupby('StoreKey')['Revenue_USD'].shift(1)

# Calculate Month-on-Month Revenue Growth Rate
monthly_revenue['MoM_Revenue_Growth'] = (
    (monthly_revenue['Revenue_USD'] - monthly_revenue['Previous_Month_Revenue']) /
    monthly_revenue['Previous_Month_Revenue']
) * 100

print('Monthly Revenue Growth Rate per Store (first few rows):')
display(monthly_revenue.head())

print('\nMonthly Revenue Growth Rate per Store (last few rows):')
display(monthly_revenue.tail())
````

````python
# Determine the current date or a reference date for age calculation
# Using the maximum order date from sales data as a reference point
reference_date = sales_full['Order Date'].max()

# Calculate store age in years
stores['Store_Age_Years'] = (reference_date - stores['Open Date']).dt.days // 365

# Define age buckets
age_bins = [0, 5, 10, 15, 20, float('inf')] # Use float('inf') to ensure the last bin is always the largest and monotonically increasing
age_labels = ['0-5 years', '5-10 years', '10-15 years', '15-20 years', '20+ years']

# Create a new column for Store Age Bucket
stores['Store_Age_Bucket'] = pd.cut(stores['Store_Age_Years'], bins=age_bins, labels=age_labels, right=False)

print('Stores DataFrame with Age and Age Bucket (first few rows):')
display(stores[['StoreKey', 'Open Date', 'Store_Age_Years', 'Store_Age_Bucket']].head())
````

````python
# Merge store age buckets with the sales_full DataFrame
sales_full_with_age = sales_full.merge(
    stores[['StoreKey', 'Store_Age_Bucket']],
    on='StoreKey',
    how='left'
)

# Calculate total revenue in each age bucket
total_revenue_by_age_bucket = sales_full_with_age.groupby('Store_Age_Bucket')['Revenue_USD'].sum().sort_values(ascending=False)

print('\nTotal Revenue by Store Age Bucket (USD):')
display(total_revenue_by_age_bucket)
````

````python
# Calculate the number of stores in each age bucket
num_stores_by_age_bucket = stores.groupby('Store_Age_Bucket')['StoreKey'].nunique()

# Calculate the total revenue for each bucket (re-calculated for clarity if previous cell not run)
total_revenue_by_age_bucket_for_avg = sales_full_with_age.groupby('Store_Age_Bucket')['Revenue_USD'].sum()

# Calculate average revenue per store in each bucket
average_revenue_per_store_by_age_bucket = (total_revenue_by_age_bucket_for_avg / num_stores_by_age_bucket).sort_values(ascending=False)

print('\nAverage Revenue per Store by Age Bucket (USD):')
display(average_revenue_per_store_by_age_bucket)
````
